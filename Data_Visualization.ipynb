```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

# Data Visualization on Titanic Dataset

## CodeAlpha Internship - Task 3

```python
sns.countplot(x="Survived", data=df)
plt.title("Survival Count")
plt.show()
```

More passengers did not survive than survived.

```python
sns.countplot(x="Sex", data=df)
plt.title("Gender Distribution")
plt.show()
```

Male passengers are more numerous than female passengers.

```python
sns.countplot(x="Pclass", data=df)
plt.title("Passenger Class Distribution")
plt.show()
```

Most passengers traveled in third class.

```python
sns.histplot(df["Age"], bins=20)
plt.title("Age Distribution")
plt.show()
```

Most passengers were between 20 and 40 years old.

```python
sns.countplot(x="Sex", hue="Survived", data=df)
plt.title("Survival by Gender")
plt.show()
```

Female passengers had a higher survival rate than male passengers.

```python
sns.boxplot(x=df["Fare"])
plt.title("Fare Distribution")
plt.show()
```

The Fare column contains some high-value outliers.

```python
numeric_df = df.select_dtypes(include="number")

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

The heatmap shows the relationships between numerical variables.
